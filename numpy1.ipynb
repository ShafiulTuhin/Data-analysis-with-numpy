```python
#1d array:
import numpy as np
arr1  = np.array([1,2,3])
print(arr1.ndim)
```

```python
#2d array:
arr2  = np.array([[1,2,3],[4,5,6]])
print(arr2.ndim)
```

```python
#3d array:
arr3 = np.array([[[1,2,3],[4,5,6]],[[1,2,3],[4,5,6]]])
print(arr3.ndim)
```

```python
##Array attribute
print(arr1.shape)
print(arr2.shape)
print(arr3.shape)

print(arr1.dtype)
print(arr2.dtype)
print(arr3.dtype)

print(arr1.size)
print(arr2.size)
print(arr3.size)
```

```python
#Set or change a data type
arr = np.array([1,2,3])
print(arr.dtype)
arr = np.array([1,2,3],dtype = np.int16)# changing data type using dtype.
print(arr.dtype)
result = arr.astype(np.int8)# changing data type using astype method.
print(result.dtype)
```

```python
#tuple
tup = (10,20,30)
arr = np.array(tup, dtype = np.int8)

print(arr)
print(type(arr))
print(arr.dtype)
```

```python
#Set
set = {10,20,30}
arr = np.array(set)
#arr = np.array(set, dtype = np.int16) Not possible to change data type here.
                                    # Need to conver to string first

arr = np.array(list(set), dtype = np.int32)
print(arr)
print(type(arr))
print(arr.dtype)
```

```python
#Dictionary
dic = {'a':10,'b':20,'c':30}

keys = dic.keys()
print(keys)
values = dic.values()
print(values)
items = dic.items()
print(items)
#Keys to ndarray
arr = np.array(list(keys))
print(arr)
print(arr.dtype)
print(arr.shape)
print(arr.size)
print(arr.ndim)

# values to ndarray
arr1 = np.array(list(values))
print(arr1)
print(arr1.dtype)
print(arr1.shape)
print(arr1.size)
print(arr1.ndim)

# items to ndarray
arr2 = np.array(list(items))
print(arr2)
print(arr2.dtype)
print(arr2.shape)
print(arr2.size)
print(arr2.ndim)
```

```python
#Array generations with random values
arr = np.random.rand(2,3) #2 rows 3 column. Generates value from 0-1
print(arr)
print(arr.dtype)

#for integer value
#np.random.randint(start,end,(shape))
arr1 = np.random.randint(1,10,(2,3)) #2 rows 3 column. Generates value from 1-10
print(arr1)
print(arr1.dtype)
print(arr1.ndim)

#for float value
#np.random.uniform(start,end,(shape))
arr2 = np.random.uniform(50,100,(3,4)) #3 rows 4 column. Generates value from 50-100
print(arr2)
print(arr2.dtype)
print(arr2.ndim)
```

```python
## ndarray with range function
#np.arrange(start,end,step)
arr = np.arange(1,10,1)
print(arr)
print(arr.dtype)
print(arr.shape)
print(arr.ndim)
print(arr.size)
print(arr.reshape(3,3))

#linespace
#np.linespace(start,end(included),koita)
arr1 = np.linspace(1,10,15)
print(arr1)
print(arr1.dtype)
print(arr1.shape)
print(arr1.ndim)
print(arr1.size)

#logspace
#np.logspace(start,end(included),koita)
arr2 = np.logspace(0,4,6,base = 2)
print("Logspace")
print(arr2)
print(arr2.dtype)
print(arr2.shape)
print(arr2.ndim)
print(arr2.size)
```

```python
## Slicing and indexing
#Indexing
arr1 = [1,2,3,4,5,6]
print(arr1[2])
arr2 = np.array([[1,2,3],
                 [4,5,6]])
print(arr2[0][2])
print(arr2[1][2])

#Slicing
print(arr1[0:3])#1d
#2d slciing
#arr[row_start:row_end:Step, coloumn_start:column:end:step]

print(arr2[0:1,])
print(arr2[1:,])
column_0 = arr2[:, 0:1]
print(column_0)
```

```python
## Copy,Advanced indexing and iteration
#Copy
arr1 = np.array([1,2,3,4,5])
arr1_mod = arr1[1:4] 
print(arr1_mod)
arr1_mod[1] = 100# changed the main array also(arr1)
print(arr1_mod)
print(arr1)
#main array will still same if use copy method
print("** Result after use copy method")
arr1_mod_copy = arr1[1:4].copy() 
print(arr1_mod_copy)
arr1_mod_copy[1] = 3# main array can't change.(arr1)
print(arr1_mod_copy)
print(arr1)
```

```python
#Advance indexing
lst = np.array([10,20,30,40,50])
values = lst[[0,4,2,3,1]] # Can find all element by index number
print(values)

arr2 = np.array([[1,2,3],
                 [4,5,6]])
print(arr2)
print(arr2[[0,1],[1,2]])
print(arr2[arr2 > 2])
```

```python
#Iteration
arr = np.array([[10,20,30],[100,200,300]])

# for i in arr:
#     for j in i:
#         print(j)

for i in np.nditer(arr):
    print(i)
```
